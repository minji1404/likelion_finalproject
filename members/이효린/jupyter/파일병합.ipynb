```python
import pandas as pd
import glob
import os

# 1. '신용정보'가 포함된 파일만 선택
file_paths = glob.glob("C:/Users/HR/Desktop/workspace/파이널프로젝트/Train_명세서/*xlsx")

# 2. 병합할 데이터프레임 리스트 생성
merged_list = []

for file in file_paths:
    try:
        df = pd.read_excel(file, sheet_name=0)

        if df.shape[0] > 0 and df.shape[1] > 0:
            기준년월 = os.path.basename(file).split("_")[0]
            df.insert(0, '기준년월', 기준년월)
            merged_list.append(df)
        else:
            print(f"⚠️ 데이터 없음: {file}")
    except Exception as e:
        print(f"❌ 에러 발생: {file} / {e}")

# 3. 병합 및 저장
if merged_list:
    merged_df = pd.concat(merged_list, ignore_index=True)
    output_path = "C:/Users/HR/Desktop/workspace/파이널프로젝트/회원정보_전체_명세서.xlsx"
    merged_df.to_excel(output_path, index=False)
    print(f"🎉 병합 완료! 저장 위치: {output_path}")
else:
    print("🚫 병합할 신용정보 파일이 없습니다.")
```

```python
import pandas as pd
import glob
import os

# 1. 병합 대상 폴더 지정
folder_path = "C:/Users/HR/Desktop/workspace/파이널프로젝트/데이터명세서"

# 2. 마케팅 관련 엑셀 파일만 찾기
file_paths = glob.glob(os.path.join(folder_path, "*신용정보*.xlsx"))

# 3. 병합 리스트 초기화
merged_list = []

for file in file_paths:
    try:
        df = pd.read_excel(file)
        # 파일명에서 출처(T train / T test) 구분
        구분 = os.path.basename(file).split("_")[0]  # 'Train' 또는 'Test'
        df.insert(0, '구분', 구분)
        merged_list.append(df)
    except Exception as e:
        print(f"❌ 에러 발생: {file} / {e}")

# 4. 병합 및 저장
if merged_list:
    merged_df = pd.concat(merged_list, ignore_index=True)
    output_path = os.path.join(folder_path, "신용정보_명세서_전체_병합.xlsx")
    merged_df.to_excel(output_path, index=False)
    print(f"✅ 병합 완료! 저장 위치: {output_path}")
else:
    print("🚫 병합할 마케팅 파일이 없습니다.")
```
