```python
# 기본
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
import pyarrow.parquet as pq
import gc
```

# 데이터 불러와서 합치기

```python
#201807

# 1. 데이터 불러오기
member_df = pd.read_parquet("train/1.회원정보/201807_train_회원정보.parquet")
sales_df = pd.read_parquet("train/3.승인매출정보/201807_train_승인매출정보.parquet")

# 2. 199번째 이후 컬럼 + ID 선택
selected_cols = sales_df.columns[199:]
sales_selected = sales_df[['ID'] + list(selected_cols)]

# 3. 병합 (Segment 컬럼 포함)
merged_df = pd.merge(member_df[['ID', 'Segment']], sales_selected, on='ID', how='inner')

# 4. 범주형 컬럼만 Label Encoding (수치형 결측치는 그대로 둠)
for col in merged_df.columns:
    if merged_df[col].dtype == 'object':
        merged_df[col] = merged_df[col].fillna("missing")
        le = LabelEncoder()
        merged_df[col] = le.fit_transform(merged_df[col].astype(str))

# 5. 상관계수 계산 (결측치 있는 행은 자동 제외되도록)
correlations = merged_df.corr(method='pearson')  # 기본값이지만 명시적으로 표시
# corr()는 pairwise 방식 → 해당 두 컬럼에 결측치가 있으면 해당 행만 제외하고 계산

# 6. CSV 저장
correlations.to_csv("corr_201807.csv", encoding='utf-8-sig')
```

```python
visualization_df = pd.read_csv("corr_201807.csv", index_col=0)

# 한글 폰트 설정
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

# Segment 기준 상위 20개 변수 추출
top_vars = visualization_df['Segment'].abs().sort_values(ascending=False).head(20).index

# 히트맵 그리기
plt.figure(figsize=(12, 10))
sns.heatmap(visualization_df.loc[top_vars, top_vars], annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Segment와 상관 높은 변수 히트맵 (201812)")
plt.tight_layout()
plt.show()
```
