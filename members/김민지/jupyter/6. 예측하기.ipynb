```python
with open(best_model_path, 'rb') as fp :
    model1 = pickle.load(fp)
    encoder1 = pickle.load(fp)
    scaler1 = pickle.load(fp)

display(model1)
display(encoder1)
display(scaler1)
```

```python
# 예측할 데이터를 읽어온다.
df1 = pd.read_csv(prediction_path)
df1
```

```python
train_df = pd.read_csv("/content/drive/MyDrive/train_pca.csv")
# feature_columns = train_df.drop(columns=['ID', 'Segment', '기준년월']).columns.tolist() wide에는 기준년월 없음
feature_columns = train_df.drop(columns=['ID', 'Segment']).columns.tolist()


# 1. ID, 기준년월 제거
#X_pred = df1.drop(columns=['ID', '기준년월'])
X_pred = df1.drop(columns=['ID'])

# 2. object형 컬럼 라벨 인코딩 (주의: fit_transform은 임시방편임)
for col in X_pred.select_dtypes(include='object').columns:
    le = LabelEncoder()
    X_pred[col] = le.fit_transform(X_pred[col].astype(str))

# 3. 학습 데이터와 동일한 컬럼 순서로 정렬
X_pred = X_pred[feature_columns]
```

```python
# 표준화 적용
X_pred_scaled = scaler1.transform(X_pred)
X_pred_scaled
```

```python
# 예측
pred = model1.predict(X_pred_scaled)
pred_labels = encoder1.inverse_transform(pred)
```

```python
# 결과 저장
df_result = pd.DataFrame({
    'ID': df1['ID'],
    'Segment': pred_labels
})

df_result.to_csv('/content/drive/MyDrive/submission_xgboost_pca.csv', index=False)
print("제출 파일 저장 완료")
```
